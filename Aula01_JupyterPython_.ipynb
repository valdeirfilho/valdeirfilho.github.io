# Ferramentas de Texto


Jupyter Notebooks, são celulas onde é possível mesclar conteúdo de programação, juntamente com textos, imagens e uma série de mídias.

## Hierarquia de títulos
Assim como em html, podemos usar 6 hierarquias de texto:

#Título 1
##Título 2
###Título 3
####Título 4
#####Título 5
######Título 6

## Edição básica

Um texto pode conter palavras em:
**negrito** e
*itálico*




Um texto também pode conter citação:


> "É melhor você tentar algo, vê-lo não funcionar e aprender com isso, do que não fazer nada".
Mark Zuckerberg



Uma lista numerada de elementos:


1.   Aprender
